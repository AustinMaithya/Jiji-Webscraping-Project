```python
from bs4 import BeautifulSoup
import requests

base_url = 'https://jiji.co.ke'
url =  "https://jiji.co.ke/computer-monitors?query=monitors"
headers = { "User-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36"}



response = requests.get(url , headers = headers)
try :
    if  response.status_code == 200:
        print("Request successful")
    else :
        print("Request failed with status code:", response.status_code)
except requests.RequestException as e:
    print("An error occurred:", e)



soup = BeautifulSoup(response.content, 'html.parser')


monitors =  soup.find_all('div', class_ = "masonry-item")
print(len(monitors))
print(monitors)

```

```python

monitorlinks = []
for item in monitors:
    for link in item.find_all('a', href=True):
        monitorlinks.append(base_url + link['href'])

print(monitorlinks[1])
```

```python
test_url  =  'https://jiji.co.ke/nairobi-central/computer-monitors/gawfolk-32-inch-gaming-monitor-fKrjEY6E3czc4TE5RAKfsbaP.html?page=1&pos=4&cur_pos=4&ads_per_page=24&ads_count=5009&lid=xZ_rjTyZWH9EKdCO&indexPosition=3'
headers =  {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/150.0.0.0 Safari/537.36"}
monitor_list  = []
for link in monitorlinks:
    response  =  requests.get(link, headers = headers)
    soup = BeautifulSoup(response.content,'html.parser')


    title = soup.find('div', class_ = 'b-advert-title-inner qa-advert-title b-advert-title-inner--h1').text.strip()
    price =  soup.find('span', class_="qa-advert-price-view-value").text.strip()
    condition =  soup.find('div', itemprop="itemCondition").text.strip()
    brand =  soup.find('div',itemprop = "brand").text.strip()
    try:
        store_address =  soup.find('div',class_ = "b-store-details__address h-mb-10").text.strip()
    except:
        store_address = "not available"
    



    monitor = {
        "title": title,
        "price": price,
        "condition": condition,
        "brand" : brand,
        "store_address": store_address
    }

    monitor_list.append(monitor)
    print("Saving: ",monitor["title"])
```
